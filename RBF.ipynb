```python
import numpy as np

#  XOR Dataset

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)
y = np.array([0, 1, 1, 0], dtype=float)

print("Input X:")
print(X)

print("\nTarget y:")
print(y)
```

```python

# RBF Centers
centers = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

print("RBF Centers:")
print(centers)
```

```python

# Gaussian RBF

sigma = 0.5

beta = 1 / (2 * sigma**2)

print("Sigma =", sigma)
print("Beta  =", beta)


def gaussian_rbf(x, center, sigma):
    """
    Calculate Gaussian Radial Basis Function.

    phi(x,c) = exp(-||x-c||^2 / (2*sigma^2))
    """

    squared_distance = np.sum((x - center) ** 2)

    return np.exp(
        -squared_distance / (2 * sigma**2)
    )
```

```python

# Squared Distance Matrix

D_squared = np.zeros(
    (len(X), len(centers))
)

for i in range(len(X)):
    for j in range(len(centers)):

        D_squared[i, j] = np.sum(
            (X[i] - centers[j]) ** 2
        )

print("\nSquared Distance Matrix D^2:")
print(D_squared)
```

```python

# RBF Activation Matrix

Phi = np.exp(-beta * D_squared)

print("\nRBF Activation Matrix Phi:")

np.set_printoptions(precision=6, suppress=True)

print(Phi)
```

```python

#  Calculate Output Weights

weights = np.linalg.solve(Phi, y)

print("\nOutput Weights:")
print(weights)
```

```python

# 7. Calculate Predictions

y_pred = Phi @ weights

print("\nPredicted Output:")
print(y_pred)
```
